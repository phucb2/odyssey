```python
# Tiny DisplayHandle check (should replace in place, not append)
from rich.panel import Panel
from rich.text import Text
import time

nb = NotebookDisplay()
for i in range(5):
    nb.update(Panel(Text(f"tick {i+1}/5"), border_style="cyan"))
    time.sleep(0.3)
print("If you only see one panel above ending at 5/5, DisplayHandle works.")
```
