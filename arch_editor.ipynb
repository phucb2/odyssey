# Architecture editor

Interactive node canvas for viewing and designing Odyssey model architectures.

**When:** exploring CNN / ResNet / DETR structure, or sketching a new Sequential net in Jupyter.

Run all cells. Requires `uv sync` (installs `anywidget`, `ipywidgets`).

```python
from odyssey.models import create_cnn_model, create_res_model
from odyssey.models.detr import create_detr_model
from odyssey.viz import ArchEditor
import torch
```

## View CNN

```python
cnn_editor = ArchEditor.from_module(create_cnn_model(), input_shape=(1, 784))
cnn_editor
```

## View ResNet

```python
res_editor = ArchEditor.from_module(create_res_model(wide=False), input_shape=(1, 784))
res_editor
```

## View DETR (coarse graph)

```python
detr_editor = ArchEditor.from_module(create_detr_model())
detr_editor
```

## Build from edited graph

After editing the CNN graph in the widget, compile and run a forward pass.

```python
model = cnn_editor.build(validate=False)
x = torch.zeros(2, 784)
model(x).shape
```
